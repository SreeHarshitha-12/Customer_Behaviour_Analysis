```python
from pathlib import Path

import numpy as np
import pandas as pd

```

```python
file_name = "../dataset/ecommerce_cleaned.csv"
df = pd.read_csv(file_name)

df.head()
```

```python
gender_map = {
    "Male": 0,
    "Female": 1,
    "Other": 2
}

df["customer_gender"] = df["customer_gender"].map(gender_map)

df.head(20)
```

```python

df.isnull().sum()

```

```python
return_map = {
    "Yes": 1,
    "No": 0
}
df["returned"] = df["returned"].map(return_map)
df.head()

```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_month"] = df["order_date"].dt.month
df["order_year"] = df["order_date"].dt.year

df["order_day_of_week"] = df["order_date"].dt.dayofweek
df.head()
```

```python
print("total_unique_rows",len(df))
print("total_number_of_customer_id",len(df["customer_id"].unique()))
print("total_number_of_order_id",len(df["order_id"].unique()))
```

```python
df.info()
```

```python
df["age_group"] = pd.cut(
    df["customer_age"],
    bins=[0, 25, 40, 60, 100],
    labels=["Young", "Adult", "Middle", "Senior"]
)
df.head()
```

```python
df.to_csv("./dataset/ecommerce_cleaned.csv", index=False)
print("Saved!")
```
