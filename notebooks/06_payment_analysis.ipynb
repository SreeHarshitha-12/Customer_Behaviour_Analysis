```python

import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")
import warnings
warnings.filterwarnings("ignore")
```

```python
file_name = "../dataset/ecommerce_cleaned.csv"
df = pd.read_csv(file_name)

df.head()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x="payment_method", palette="Blues_d")
plt.title("Most Used Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
payment_counts = df["payment_method"].value_counts()

plt.figure(figsize=(6, 6))

colors = ["skyblue", "orange", "lightgreen", "pink", "violet", "gold"]

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    colors=colors
)

plt.title("Payment Method Distribution")
plt.show()
```

```python
# Average order value by payment method
aov_by_payment = (
    df.groupby("payment_method")["total_amount"]
      .mean()
      .sort_values(ascending=False)
      .reset_index(name="avg_order_value")
)

plt.figure(figsize=(8, 4))
sns.barplot(data=aov_by_payment, x="payment_method", y="avg_order_value", palette="Blues_d")
plt.title("Average Order Value by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Average Order Value")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

aov_by_payment
```

```python
# Discount behavior by payment method
discount_by_payment = (
    df.groupby("payment_method")["discount"]
      .mean()
      .sort_values(ascending=False)
      .reset_index(name="avg_discount")
)

plt.figure(figsize=(8, 4))
sns.barplot(data=discount_by_payment, x="payment_method", y="avg_discount", palette="Greens_d")
plt.title("Average Discount by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Average Discount")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

discount_by_payment
```

```python
# Return rate by payment method
return_rate_by_payment = (
    df.groupby("payment_method")["returned"]
      .mean()
      .sort_values(ascending=False)
      .reset_index(name="return_rate")
)

plt.figure(figsize=(8, 4))
sns.barplot(data=return_rate_by_payment, x="payment_method", y="return_rate", palette="Reds_d")
plt.title("Return Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Return Rate")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

return_rate_by_payment
```

```python
# Profit margin by payment method
profit_by_payment = (
    df.groupby("payment_method")["profit_margin"]
      .mean()
      .sort_values(ascending=False)
      .reset_index(name="avg_profit_margin")
)

plt.figure(figsize=(8, 4))
sns.barplot(data=profit_by_payment, x="payment_method", y="avg_profit_margin", palette="Purples_d")
plt.title("Average Profit Margin by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Average Profit Margin")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

profit_by_payment
```

```python
# Payment method by region
region_payment = pd.crosstab(df["region"], df["payment_method"])

plt.figure(figsize=(10, 5))
sns.heatmap(region_payment, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Payment Method by Region")
plt.xlabel("Payment Method")
plt.ylabel("Region")
plt.tight_layout()
plt.show()

region_payment
```
