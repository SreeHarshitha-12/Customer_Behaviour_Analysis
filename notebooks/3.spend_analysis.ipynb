```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
sns.set_theme(style="whitegrid")
```

```python
file_name = "../dataset/ecommerce_cleaned.csv"
df = pd.read_csv(file_name)

df.head()
```

```python
print(df["total_amount"].describe())
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df["customer_age"], bins=20, kde=True, color="#264653")
plt.title("Distribution of Customer Age")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")
plt.show()
```

```python
age_spend = df.groupby("age_group", observed=True)["total_amount"].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=age_spend, x="age_group", y="total_amount", palette="Blues_d")
plt.title("Average Spend by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Avg Total Amount")
plt.tight_layout()
plt.show()
```

```python
gender_spend = (
    df.groupby("customer_gender", observed=True)["total_amount"]
    .mean()
    .reset_index()
)

gender_spend["customer_gender"] = gender_spend["customer_gender"].replace({
    0: "Male",
    1: "Female",
    2: "Other"
})

plt.figure(figsize=(6, 4))

sns.barplot(
    data=gender_spend,
    x="customer_gender",
    y="total_amount",
    palette=["#E7BE5E", "#FF8BC5", "#C68DFA"]
)

plt.title("Average Spend by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Total Amount")

plt.tight_layout()
plt.show()
```

```python
region_spend = df.groupby("region")["total_amount"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=region_spend, x="region", y="total_amount", palette="Greens_d")
plt.title("Average Spend by Region")
plt.xlabel("Region")
plt.ylabel("Avg Total Amount")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="age_group", y="total_amount", palette="Blues", order=["Young", "Adult", "Middle", "Senior"])
plt.title("Spend Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Total Amount")
plt.ylim(0, 1500)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="region", y="total_amount", palette="Greens")
plt.title("Spend Distribution by Region")
plt.xlabel("Region")
plt.ylabel("Total Amount")
plt.ylim(0, 1500)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="payment_method", palette="Blues_d")
plt.title("Most Used Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
