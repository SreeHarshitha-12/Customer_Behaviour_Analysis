```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
sns.set_theme(style="whitegrid")
import warnings
warnings.filterwarnings("ignore")
```

# Customer Demographic Analysis


```python
file_name = "../dataset/ecommerce_cleaned.csv"
df = pd.read_csv(file_name)

df.head()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x="customer_gender", palette="Set2")
plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x="region", palette="Greens")
plt.title("Customer Distribution by Region")
plt.xlabel("Region")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x="age_group", palette="Blues_d")
plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
##custoer analysis:-
sns.histplot(df['customer_age'], bins=20, kde=True)
plt.title("Customer Age Distribution")
plt.show()
```

```python
sns.countplot(x='customer_gender', data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
customer_spending = df.groupby('customer_id')['total_amount'].sum()

sns.histplot(customer_spending, bins=30)
plt.title("Customer Spending Distribution")
plt.show()
```
