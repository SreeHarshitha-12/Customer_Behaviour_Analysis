```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
sns.set_theme(style="whitegrid")
```

```python
file_name = "../dataset/ecommerce_cleaned.csv"
df = pd.read_csv(file_name)

df.head()
```

```python
category_sales = df.groupby("category")["quantity"].sum().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=category_sales, x="category", y="quantity", palette="Blues_d")
plt.title("Most Sold Category")
plt.xlabel("Category")
plt.ylabel("Total Quantity Sold")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(4, 4))

plt.pie(
    category_sales["quantity"],
    labels=category_sales["category"]
)

plt.title("Category Sales Share")
plt.show()
```

```python
category_profit = df.groupby("category")["profit_margin"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=category_profit, x="category", y="profit_margin", palette="Greens_d")
plt.title("Average Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Avg Profit Margin")
plt.tight_layout()
plt.show()
```

```python
category_discount = df.groupby("category")["discount"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=category_discount, x="category", y="discount", palette="Oranges_d")
plt.title("Average Discount by Category")
plt.xlabel("Category")
plt.ylabel("Avg Discount")
plt.tight_layout()
plt.show()
```

```python
category_revenue = df.groupby("category")["total_amount"].sum().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=category_revenue, x="category", y="total_amount", palette="Oranges_d")
plt.title("Total Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.tight_layout()
plt.show()
```

```python
## 
df['category'].value_counts().plot(kind='bar')
plt.title("Product Category Count")
plt.show()
```

```python
df.groupby('category')['quantity'].sum().sort_values(ascending=False).plot(kind='bar')
plt.title("Most Sold Categories")
plt.show()
```

```python
df.groupby('category')['profit_margin'].sum().sort_values(ascending=False).plot(kind='bar')
plt.title("Most Profitable Categories")
plt.show()
```
