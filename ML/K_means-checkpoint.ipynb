```python
# Step 1 — Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

_ = (KMeans, StandardScaler)
```

```python
# Step 2 — Load RFM dataset
rfm = pd.read_csv('../dataset/rfm_analysis.csv')
rfm.head()
```

```python
X = rfm[['recency', 'frequency', 'monetary']]
X.columns = ['Recency', 'Frequency', 'Monetary']

X.head()
```

```python
# Step 4 — Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

```python
# Step 5 — Find optimal clusters using Elbow Method
wcss = []
K_range = range(1, 11)

for k in K_range:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    wcss.append(model.inertia_)

print("WCSS computed for k=1..10")
```

```python
# Step 6 — Plot Elbow Graph
plt.figure(figsize=(8,4))
plt.plot(list(K_range), wcss, marker='o')
plt.xticks(list(K_range))
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()
```

```python
# Step 7 — Apply K-Means (choose k based on elbow; default 4)
k = 4  # change after inspecting the elbow plot
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
rfm['Cluster'] = kmeans.fit_predict(X_scaled)

print(rfm['Cluster'].value_counts().sort_index())
```

```python
# Step 8 — Analyze cluster characteristics
cluster_summary = rfm.groupby('Cluster')[['recency', 'frequency', 'monetary']].mean()
display(rfm.head())
display(cluster_summary)

# Step 9 — Add simple customer segment names
segment_map = {
    0: 'Budget Customers',
    1: 'Regular Customers',
    2: 'Premium Customers',
    3: 'Inactive Customers',
}

rfm['Customer_Segment'] = rfm['Cluster'].map(segment_map)
display(rfm[['recency', 'frequency', 'monetary', 'Cluster', 'Customer_Segment']].head())
```

```python
# Cluster distribution
plt.figure(figsize=(8,4))
sns.countplot(data=rfm, x='Customer_Segment', palette='Set2', order=rfm['Customer_Segment'].value_counts().index)
plt.title('Customer Segment Distribution')
plt.xlabel('Customer Segment')
plt.ylabel('Customer Count')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Additional visuals: boxplot and heatmap of cluster means
plt.figure(figsize=(8,5))
sns.boxplot(x='Cluster', y='monetary', data=rfm, palette='Pastel1')
plt.title('Monetary by Cluster')
plt.xlabel('Cluster')
plt.ylabel('Monetary')
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(cluster_summary, annot=True, cmap='YlGnBu', fmt='.2f')
plt.title('Cluster Averages')
plt.tight_layout()
plt.show()
```
